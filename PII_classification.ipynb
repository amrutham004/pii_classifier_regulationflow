```python
#Checking whether the patient_id from all datasets exists in the patient dataset
patient_ids = set(patients["id"])

for name, df in datasets.items():
    if "patient_id" in df.columns:
        invalid_ids = (~df["patient_id"].isin(patient_ids)).sum()

        print(name, "Patient IDs not found in Patients:", invalid_ids)
```

```python
#Checking for appointment_id in Billing and Lab Reports
appointment_ids = set(appointments["id"])

for name, df in datasets.items():
    if "appointment_id" in df.columns:
        invalid_ids = (
            ~df["appointment_id"].isin(appointment_ids)
        ).sum()

        print(
            name,
            "Appointment IDs not found in Appointments:",
            invalid_ids
        )
```

```python
#Checking Categorical value inconsistency 
for name, df in datasets.items():
    print(f"\nDataset: {name}")

    for column in df.select_dtypes(include=["object", "str"]).columns:
        print(f"\n{column}")
        print(df[column].value_counts().head(15))
```

```python
#Detecting inconsistent capitalization and whitespace
for name, df in datasets.items():
    print(f"\nDataset: {name}")

    for column in df.select_dtypes(
        include=["object", "str"]
    ).columns:

        values = df[column].dropna().astype(str)

        inconsistent_case = (
            values.str.lower().nunique()
            != values.nunique()
        )

        if inconsistent_case:
            print(f"Possible capitalization inconsistency: {column}")
```

```python
#Checking leading and trailing spaces
for name, df in datasets.items():
    print(f"\nDataset: {name}")

    for column in df.select_dtypes(
        include=["object", "str"]
    ).columns:

        values = df[column].dropna().astype(str)

        leading = values.str.startswith(" ").sum()
        trailing = values.str.endswith(" ").sum()

        if leading > 0 or trailing > 0:
            print(
                column,
                "Leading spaces:", leading,
                "Trailing spaces:", trailing
            )
```

```python
#Checking for empty strings
for name, df in datasets.items():
    print(f"\nDataset: {name}")

    for column in df.select_dtypes(
        include=["object", "str"]
    ).columns:

        empty_count = (
            df[column].fillna("").astype(str).str.strip() == ""
        ).sum()

        if empty_count > 0:
            print(column, "Empty strings:", empty_count)
```

```python
df = appointments["follow_up_notes"]

print("NaN values:", df.isna().sum())

print(
    "Empty strings:",
    (df.notna() & df.astype(str).str.strip().eq("")).sum()
)

print("Total missing or empty:", df.isna().sum() +
      (df.notna() & df.astype(str).str.strip().eq("")).sum())
```

```python
#Validating date columns
date_columns = {
    "Appointments": ["appointment_date", "created_at"],
    "Billing": ["payment_date"],
    "Employee Records": ["date_of_birth", "joined_at"],
    "Lab Reports": ["collected_at", "reported_at"],
    "Patients": ["date_of_birth", "registered_at"]
}

for name, columns in date_columns.items():
    df = datasets[name]

    print(f"\nDataset: {name}")

    for column in columns:
        converted = pd.to_datetime(df[column], errors="coerce")

        invalid_count = (
            df[column].notna() & converted.isna()
        ).sum()

        print(column, "Invalid dates:", invalid_count)
```

```python
#Checking future dates of birth
today = pd.Timestamp.today().normalize()

for name in ["Employee Records", "Patients"]:
    df = datasets[name]

    dob = pd.to_datetime(
        df["date_of_birth"],
        errors="coerce"
    )

    future_dob = (dob > today).sum()

    print(name, "Future dates of birth:", future_dob)
```

```python
#Checking logical date relationships
# Check appointment dates against creation dates
df = appointments.copy()

appointment_date = pd.to_datetime(df["appointment_date"])
created_date = pd.to_datetime(df["created_at"])

print(
    "Appointments created before appointment date:",
    (created_date.dt.date < appointment_date.dt.date).sum()
)

# Check lab report dates
df = lab_reports.copy()

collected = pd.to_datetime(df["collected_at"])
reported = pd.to_datetime(df["reported_at"])

print(
    "Reports before collection:",
    (reported < collected).sum()
)

# Check future employee joining dates
df = employee_records.copy()

joined = pd.to_datetime(df["joined_at"])

print(
    "Future employee joining dates:",
    (joined > today).sum()
)
```

```python
#Remaining Categorical checks
# Complete categorical profiling across all datasets

categorical_columns = {
    "Appointments": [
        "status",
        "specialisation",
        "time_slot",
        "doctor_name"
    ],
    "Billing": [
        "payment_mode",
        "is_refunded"
    ],
    "Employee Records": [
        "department",
        "designation"
    ],
    "Lab Reports": [
        "test_name",
        "interpretation"
    ],
    "Patients": [
        "gender",
        "blood_group",
        "insurance_provider",
        "is_active"
    ]
}

for name, columns in categorical_columns.items():

    df = datasets[name]

    print("\n" + "=" * 60)
    print(f"DATASET: {name}")
    print("=" * 60)

    for column in columns:

        print(f"\nColumn: {column}")
        print("Unique values:", df[column].nunique(dropna=True))
        print("Missing values:", df[column].isna().sum())
        print("Distinct categories and counts:")

        print(df[column].value_counts(dropna=False))
```

```python
#Reviewing identifier data types
# Identify identifier-like columns and their current data types

identifier_keywords = [
    "id",
    "number",
    "pincode",
    "ifsc",
    "pan"
]

for name, df in datasets.items():

    print(f"\n{'=' * 50}")
    print("Dataset:", name)
    print("=" * 50)

    for column in df.columns:

        if any(
            keyword in column.lower()
            for keyword in identifier_keywords
        ):
            print(
                column,
                "| Data type:", df[column].dtype
            )
```

```python
# Create independent working copies

cleaned_datasets = {
    name: df.copy()
    for name, df in datasets.items()
}

# Columns that should be represented as strings

columns_to_convert = {
    "Billing": [
        "bank_account_number"
    ],

    "Employee Records": [
        "mobile_number",
        "aadhaar_number",
        "bank_account_number"
    ],

    "Patients": [
        "mobile_number",
        "aadhaar_number",
        "pincode"
    ]
}

# Convert identifier columns

for name, columns in columns_to_convert.items():

    for column in columns:

        cleaned_datasets[name][column] = (
            cleaned_datasets[name][column].astype("string")
        )

print("Identifier conversion completed.")
```

```python
for name, columns in columns_to_convert.items():

    print(f"\nDataset: {name}")

    for column in columns:

        print(
            column,
            "| Original:",
            datasets[name][column].dtype,
            "| Cleaned:",
            cleaned_datasets[name][column].dtype
        )
```

```python
#Verifying data integrity after conversion
# Final validation after identifier conversion

for name, columns in columns_to_convert.items():

    original_df = datasets[name]
    cleaned_df = cleaned_datasets[name]

    print(f"\nDataset: {name}")

    for column in columns:

        print(f"\nColumn: {column}")

        print(
            "Original record count:",
            len(original_df[column])
        )

        print(
            "Cleaned record count:",
            len(cleaned_df[column])
        )

        print(
            "Original missing values:",
            original_df[column].isna().sum()
        )

        print(
            "Cleaned missing values:",
            cleaned_df[column].isna().sum()
        )

        print(
            "Unique values preserved:",
            original_df[column].nunique()
            == cleaned_df[column].nunique()
        )
```

```python
#Comparing all five original and cleaned datasets
for name in datasets:

    original = datasets[name]
    cleaned = cleaned_datasets[name]

    print(f"\n{'=' * 45}")
    print("Dataset:", name)
    print("=" * 45)

    print("Original shape:", original.shape)
    print("Cleaned shape:", cleaned.shape)

    print(
        "Row count preserved:",
        len(original) == len(cleaned)
    )

    print(
        "Column count preserved:",
        len(original.columns) == len(cleaned.columns)
    )

    print(
        "Missing values before:",
        original.isna().sum().sum()
    )

    print(
        "Missing values after:",
        cleaned.isna().sum().sum()
    )

    print(
        "Duplicate rows after cleaning:",
        cleaned.duplicated().sum()
    )
```
